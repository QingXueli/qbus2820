```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

spend_years = pd.read_excel("AmountSpent_13to17.xlsx")
spend_years.head()
```

```python
# Alternative 1
plt.plot(spend_years.mean()[1:6])
plt.xlabel("Year")
plt.ylabel("Average Customer Spend")
plt.title("Historical Average Customer Spending")

plt.show()
```

```python
# Alternative 2
fig = plt.figure()

spend_years = spend_years.iloc[:,1:]
plt.plot([1,2,3,4,5], spend_years.mean())
plt.xticks( np.arange(1,6), ['2013', '2014', '2015', '2016', '2017'] )

plt.xlabel("Year")
plt.ylabel("Average Customer Spend")
plt.title("Historical Average Customer Spending")

#fig
```
